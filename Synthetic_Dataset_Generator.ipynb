```python
import pandas as pd
import numpy as np
import os
from datetime import datetime, timedelta

def generate_synthetic_data(num_rows=10000):
    np.random.seed(42)
    gateways = ["razorpay", "payu", "stripe"]
    statuses = ["SUCCESS", "FAILED"]
    error_reasons = ["insufficient_funds", "gateway_timeout", "fraud_suspected", "gateway_error", None]

    start_time = datetime.now() - timedelta(days=7)

    data = {
        "txn_id": [f"TXN_{i:06d}" for i in range(num_rows)],
        "amount": np.round(np.random.exponential(scale=2500, size=num_rows) + 100, 2),
        "gateway": np.random.choice(gateways, size=num_rows, p=[0.5, 0.3, 0.2]),
        "status": np.random.choice(statuses, size=num_rows, p=[0.81, 0.19]),
        "retry_count": np.random.poisson(lam=0.3, size=num_rows),
        "ts": [start_time + timedelta(seconds=int(np.random.randint(0, 604800))) for _ in range(num_rows)]
    }

    df = pd.DataFrame(data)
    # Assign error reasons only to failed transactions
    df["error_reason"] = df["status"].apply(lambda s: np.random.choice(error_reasons[:4]) if s == "FAILED" else None)

    # Ensure the directory exists before saving
    os.makedirs("data/sample", exist_ok=True)

    df.to_csv("data/sample/synthetic_payments.csv", index=False)
    print(f"Generated {num_rows} synthetic records successfully!")

if __name__ == "__main__":
    generate_synthetic_data()
```
